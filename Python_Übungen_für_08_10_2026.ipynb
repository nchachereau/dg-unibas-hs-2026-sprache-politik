### 2.2 Funktionen und Variablen

Die wichtigsten Bausteine eines Computerprogramms sind **Funktionen** und **Variablen**.

**Funktionen** sind wie kleine Maschinen: man übergibt ihnen etwas und sie geben etwas zurück. Zum Beispiel die Funktion `len()`, die wir im Kurs schon mehrmals erwähnt haben:

````python
len("Etwas Text")
````

Hier wird die Zeichenkette `Etwas Text` der Funktion `len()` übergeben, und die Funktion gibt die Länge der Zeichenkette zurück.

Das, was einer Funktion übergeben wird, wird **Argument** oder Parameter genannt.

Um eine Funktion auszuführen / aufzurufen, schreibt man ihr Name (`len`) und unmittelbar danach runde Klammern (also `len()`), und zwischen den Klammern werden die Argumente angegeben, also die Werte, die der Funktion übergeben werden sollen. Mehrere Argumente werden mit Kommas getrennt:

```python
print("a", "b", "c")
```

Es gibt Funktionen, die nichts zurückgeben, und nur etwas tun, wie zum Beispiel `print()`. Es kann auch Funktionen geben, denen nichts übergeben werden muss, d.h. die keine Argumente annehmen.

Kommen wird zu den **Variablen**. Ich zitiere aus dem Kurs «Python für Historiker:innen»:

> Sie können sich eine Variable als eine Art Behälter (container) vorstellen, der für die Zeit des Programmdurchlaufs bestimmte Werte aufbewahrt und ansteuerbar macht.

Im folgenden Code wird der Wert (die Zahl) 42 der Variable `the_answer` zugewiesen:

````python
the_answer = 42
````

Diese Variable kann jetzt weitergenutzt werden, zum Beispiel so:

````python
print(
    "What is the Answer to the Ultimate Question of Life, the Universe, and Everything?"
)
print("The answer is:", the_answer)
````

Der Name ist `the_answer` ist willkürlich. Die Variable hätten wir auch anders nennen können und der Code hätte genau so gut funktioniert. Namen von Variablen (und von Funktionen) dienen nicht der richtigen Ausführung des Codes. Sie helfen aber, anderen Menschen (bzw. sich selber in der Zukunft!) das Verständnis des Codes zu erleichtern.

Einer Variable kann nicht nur ein festgeschriebener Wert (wie hier die Zahl 42) zugewiesen werden, aber auch das, was eine Funktion zurückgegeben hat, wie zum Beispiel hier:

````python
# Variable mit festgeschriebenem Wert
my_text = "My text"

# Variable mit dem Ergebnis einer Funktion
length_of_my_text = len(my_text)

# Zeigen wir diesen Wert an
print(length_of_my_text)
````

Es gibt auch eine spezielle Art von Funktionen, sog. **Methoden**, die einem Objekt (zum Beispiel einer Zeichenkette) sozusagen angeschlossen sind. Im Unterschied zu Funktionen wie `len()` oder `print()`, die verschiedene Datentypen als Argumente annehmen, sind Methoden nur für einen Datentyp direkt relevant.

Wenn wir kurz ein Beispiel ausserhalb der digitalen Welt auswählen, könnte man sich vorstellen, dass die Funktion `look_at()` mit allen möglichen Arten von Objekten umgehen kann, während `meow()` (also miauen), eine Methode vom Datentyp Katze ist, weil sie nur für diese Art von Daten relevant ist

Ein anderer Unterschied besteht darin, dass Methoden anders aufgerufen werden: sie folgen dem Objekt mit einem Punkt. Während man also `look_at(cat)` schreiben würde (oder `look_at(dog)`), müsste man `cat.meow()` schreiben. `meow(cat)` würde nicht gehen, weil `meow()` keine allgemein gültige Funktion wäre, sondern nur eine Methode von Katzen; und `dog.meow()` wäre ein Fehler, weil Hunde nicht miauen können…

Eine Methode kann auch Argumente haben. Vielleicht könnte die Katze auch eine bestimmte Höhe springen, etwa so: `cat.jump(2)`.

Hier folgen reale Beispiele mit Methoden von Zeichenketten, die Du im Kapitel [Arbeiten mit Strings](https://digital-history-berlin.github.io/Python-fuer-Historiker-innen/ch01-atomare-datentypen/03-datentyp-strings.html) schon gesehen hast:

````python
# hier wird die Methode `count()` direkt nach einer Zeichenkette aufgerufen
# und einer Variable zugewiesen, die wir `count_of_letter_e` nennen
count_of_letter_e = "Dies ist eine Zeichenkette".count("e")

print(count_of_letter_e)
````

````python
# hier wird eine Zeichenkette zuerste einer Variable zugewiesen
my_text = "DAS IST/WAR EINE GROSSGESCHRIEBENE ZEICHENKETTE"

# hier wird das Ergebnis der Methode `.lower()` einer neuen Variable zugewiesen
my_lower_text = my_text.lower()

print(my_text)
print(my_lower_text)
````

Wie andere Funktionen wird einer Methode etwas gegeben und sie gibt etwas zurück. Im Beispiel `my_text.lower()` wird der Methode nichts explizites gegeben, aber sie greift auf das Objekt zurück, an dem sie angeschlossen ist (also hier die Variable `my_text` bzw. ihr Inhalt). Im Beispiel `….count("e")` arbeitet die Methode sowohl mit der Zeichenkette `"Dies ist eine Zeichenkette"` sowie mit der Zeichenkette `"e"`, die ihr als Argument übergeben wird.

In beiden Fällen gibt die Methode einen neuen Wert zurück: das ursprüngliche Objekt wird in diesen Fällen nicht verändert, weil Strings nicht veränderbar sind.

### 2.4 Dateien einlesen

Um eine Datei einzulesen, muss sie zuerst mit der Funktion `open(path)` geöffnet werden. `open()` gibt ein **File-Objekt** zurück, das verschiedene Methoden anbietet. Die wichtigste für uns ist `.read()`, die den Inhalt der Datei als eine Zeichenkette zurückgibt. Schliesslich muss die Datei geschlossen werden, mit der Methode `.close()`.

Falls ein Fehler zwischen `open()` und `.close()` aufkommt, wird die Datei nicht geschlossen, was zu Problemen führen kann. Deshalb ist es immer besser, mit der Anweisung `with` zu arbeiten. Das funktioniert so:

```python
with open("file.txt") as f:
  file_content = f.read()

print(file_content)
```

Hier würde also die Datei "file.txt" geöffnet; das File-Objekt, das `open()` zurückgibt, wird in der Variable `f` gespeichert; die Methode `.read()` des File-Objekts `f` wird aufgerufen und ihr Ergebnis wird der Variable `file_content` zugewiesen. Das File-Objekt `f` wird automatisch geschlossen, sobald der Code-Block nach dem `with` fertig ist.

Wie bei anderen Blocks (etwa die Bedingung `if`) endet die Anweisung `with ...` mit einem Doppelpunkt, und die Zeilen nach dem Doppelpunkt müssen eingerückt werden (mit der Taste Tab ↹). Sobald der Code wieder auf eine tiefere Ebene landet (hier bei der Zeile `print(file_content)`), ist der Code-Block fertig und im Fall von `with` wird das File-Objekt (hier `f` genannt) geschlossen.

`open()` muss ein Pfad als Argument gegeben werden. Siehe dazu die Erklärungen im [entsprechenden Kapitel](https://digital-history-berlin.github.io/Python-fuer-Historiker-innen/ch04-dateien-verarbeiten/02-dateien-i-o.html) in «Python für Historiker:innen».

### 3.1 Einführung eines neuen Datentyps: Listen

#### 3.1.1 Definition von Listen

> Als Datentyp bilden Listen eine Vielzahl von Werten in einer geordneten Sequenz ab.

Listen werden in Python mit eckigen Klammern geschrieben:

```python
my_list = [1, 8, 10]
```

In Listen können beliebige Datentypen gespeichert werden. Zum Beispiel Zahlen, Zeichenketten oder [boolesche Werte](https://digital-history-berlin.github.io/Python-fuer-Historiker-innen/ch02-programmablaeufe-strukturieren/02-bedingte-anweisungen.html#boolesche-werte):

```python
my_other_list = ["Das", "ist", "eine", "Liste"]
yet_another_list = [True, 0.2, "Test"]
```

Eine Liste kann auch andere Listen beinhalten. Sie lassen sich beliebig tief verschachteln:

```python
my_list_of_lists = [["Erste", "untere", "Liste"], ["Zweite", "Liste"]]
```

#### 3.1.2 Listen und `for`-Schleifen

Im Kapitel [Schleifen](https://digital-history-berlin.github.io/Python-fuer-Historiker-innen/ch02-programmablaeufe-strukturieren/03-schleifen.html) wurden `for`-Schleifen eingeführt. Neben der Schreibweise mit `range()`, die dort vorgestellt wurde und es erlaubt, Zahlen aufzuzählen, können `for`-Schleifen Anweisungen für jedes Element einer Liste wiederholen.

Fangen wir einfach an: hier wird einfach jedes Element mit `print()` angezeigt:

````python
example_list = ["Element 1", "Element 2", "Element 3", "Element 4"]

for item in example_list:
  print(item)
````

Die `for`-Schleife wird also so geschrieben: dem Wort `for` folgt der Name der Variable, der bei jedem Durchlauf der Schleife das nächste Element der Liste zugewiesen wird. Dann kommt das Wort `in` und dann die Liste (entweder direkt als festgeschriebene Liste oder als Variable, die eine Liste enthält). Dann folgt wie so oft ein Doppelpunkt und ein eingerückter Codeblock ab der nächsten Zeile.

Es liest sich fast wie ein normaler Satz: «für jedes Element X in der Liste Y: mach das und das» (oder eben auf Englisch: «for each item `item` in the list `name of the list`: do this and that»).

Ein anderes Beispiel:

````python
# Eine Liste von Namen
names = ["Lea", "John", "Michel", "Hansjörg"]

# Weise jeden Namen nach dem anderen aus der List `names` der Variable `name` zu
for name in names:
  # … und zeige den grossgeschriebenen Namen an
  print(name.upper())
````

### 4.2 Erste Anpassungen

Nun möchten wir diese Suchfunktion verändern. Wir wollen nämlich, dass sie nur nach ganze Wörter suchen.

Für dieses Ziel muss eine Methode von Zeichenketten eingeführt werden, sowie ein Vergleichsoperator für Listen:

#### 4.2.1 `.split(separator)`

Die Methode `.split(separator)` einer Zeichenkette «zerteilt» diese Zeichenkette und gibt eine **Liste** zurück. Mit dem Argument `separator` wird angegeben, anhand welcher anderen Zeichenkette die Zeichenkette zerteilt werden soll. Zum Beispiel:

````python
"a,b,c".split(",")
````

Wenn kein Argument angegeben wird, wird die Zeichenkette anhand der Leerzeichen zerteilt:

````python
"(Digitale) Geschichte macht Spass.".split()
````

Somit lässt sich also ein Text in «Wörtern» zerteilen.

#### 4.2.2 `in` für Listen

Zur Erinnerung: Dank dem Wörten `in` kann geprüft werden, ob eine Zeichenkette Teil einer anderen Zeichenkette ist:

````python
if "e" in "Zeichenkette":
  print("'Zeichenkette' beinhaltet 'e'")
````

(Falls Dir diese Information in der Übung III gefehlt hat und Du sie nicht gefunden hattest, kannst Du hier zurückgehen und erneut versuchen, die Aufgabe zu lösen)

Ganz ähnlich kann das Wörtchen `in` verwendet werden, um zu prüfen, ob ein Element in einer **Liste** vorhanden ist.

````python
some_list = [0, 10, 2, 9, 18]

if 0 in some_list:
  print("0 ist in der Liste")

if 4 in some_list:
  print("4 ist in der Liste")
````
